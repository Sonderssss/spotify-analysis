```python
import pandas as pd

df = pd.read_parquet('../data/processed/spotify_history_clean.parquet')

df.head()
```

```python
# duration of time played for each platfrom
df.groupby('platform')['ms_played'].sum().sort_values(ascending=False)
```

```python
# finding the mean of skipped column since it is a boolean value it will give us the rate
df.groupby('platform')['skipped'].mean().sort_values(ascending=False)
```

```python
# skip rate by shuffle (was shuffle used to play the song)
df.groupby('shuffle')['skipped'].mean().sort_values(ascending=False)
```

```python
# rate at which songs were shuffled
# shuffled plays have a higher skip rate
df['shuffle'].value_counts(normalize=True)
```

```python
#skip rate and count by reason_start. Why user was skipping songs
df.groupby('reason_start')['skipped'].agg(['count', 'mean']).sort_values(by='count', ascending=False)
```

```python
top_20_artists = df['artist_name'].value_counts().head(20)
top_20_artists
```

```python
# create dataframe for the top 20 artists
top_20_df = df[df['artist_name'].isin(top_20_artists.index)]

top_20_df

# rate of skips for every artist
top_20_df.groupby('artist_name')['skipped'].mean().sort_values(ascending=False)
```

```python
# finding the skip rate over the years
# absurd; 0 skips between 2018 and 2021, and barely any in 2017 
df.groupby('year')['skipped'].agg(['mean','count'])
```

```python
# frequency of skips over the years
# between the years 218 and 2021 there were no skips, there is not corresponding True values and their counts
df.groupby('year')['skipped'].value_counts()
```

```python
# was the forward button being used especially between 2017 and 2021 
# trying to establish causation of 0 skips between the same years
# while also showing how using the fwd btn has been declining over the years since 2017
fwdbtn_by_year = df[df['reason_end'] == 'fwdbtn'].groupby('year')['reason_end'].count()
total_by_year = df.groupby('year').size()
fwdbtn_rate_by_year = fwdbtn_by_year / total_by_year
fwdbtn_rate_by_year
```
