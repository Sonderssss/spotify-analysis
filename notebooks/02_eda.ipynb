```python
import pandas as pd 

spotify_data = pd.read_csv('../data/spotify_history.csv')

spotify_data.head()
```

```python
spotify_data.info()
```

```python
spotify_data['ts'] = pd.to_datetime(spotify_data['ts'])
```

```python
#creating year column
spotify_data['year'] = spotify_data['ts'].dt.year

#confirming
spotify_data.head()

#group by year and count the number of plays
spotify_data.groupby('year')['ms_played'].count() 
```

```python
spotify_data.groupby('year')['ms_played'].sum()
```

```python
spotify_data.groupby('year')['ms_played'].sum() / spotify_data.groupby('year')['ms_played'].count()
```

```python
spotify_data['month'] = spotify_data['ts'].dt.month

spotify_data.head()

spotify_data.groupby('month')['ms_played'].count()
```

```python
spotify_data.groupby('month')['ms_played'].sum()
```

```python
spotify_data['hour'] = spotify_data['ts'].dt.hour

spotify_data.head()

spotify_data.groupby('hour')['ms_played'].count()
```

```python
spotify_data['day_of_week'] = spotify_data['ts'].dt.day_name()

spotify_data.head()
```

```python
spotify_data.groupby('day_of_week')['ms_played'].count()
```

```python
spotify_data.groupby('day_of_week')['ms_played'].sum()
```

```python
spotify_data['ts_localized'] = spotify_data['ts'].dt.tz_localize('UTC').dt.tz_convert('Africa/Nairobi').dt.tz_localize(None)

spotify_data.head()
```

```python
spotify_data['localized_hour'] = spotify_data['ts_localized'].dt.hour

spotify_data.groupby('localized_hour')['ms_played'].count()
```

```python
spotify_data.groupby('localized_hour')['ms_played'].sum()
```

```python
#EDA on top artists all time artist play counts
spotify_data.groupby('artist_name')['ms_played'].sum().sort_values(ascending=False).head(20)
```

```python
artist_summary = spotify_data.groupby('artist_name')['ms_played'].agg(['mean','count'])

print(artist_summary[artist_summary['count'] > 10].sort_values(by='mean', ascending=False).head(20))
```

```python
#listening over the years

spotify_data['year'].head()

print(spotify_data.groupby('year')['artist_name'].agg(['nunique']))
```

```python
#Favourite artists over the years

favourite_artists = spotify_data.groupby(['year', 'artist_name'])['artist_name'].size().to_frame('artist_count')


top_artists_idx = favourite_artists.groupby(level='year')['artist_count'].idxmax()

top_artists = favourite_artists.loc[top_artists_idx]

print(top_artists)
```

```python
#finding one hit wonders
artist_counts = spotify_data['artist_name'].value_counts()
one_hit_wonders = artist_counts[artist_counts == 1]
one_hit_wonders
```

```python
# the number of one hit wonders in the dataset
#len(one_hit_wonders) 
one_hit_wonders.shape[0]
```

```python
# number of unique artists in the dataset
no_of_unique_artists = spotify_data['artist_name'].nunique()
no_of_unique_artists 
```

```python
# percentage of one hit wonders in the dataset
percentage_one_hit_wonders = (one_hit_wonders.shape[0] / no_of_unique_artists) * 100
percentage_one_hit_wonders
```

```python
# one hit wonder tracks
track_counts = spotify_data['track_name'].value_counts()

one_hit_wonder_tracks = track_counts[track_counts == 1]

one_hit_wonder_tracks
```

```python
# number of one  hit wonder tracks in the dataset
one_hit_wonder_tracks.shape[0]
```

```python
# number of unique tracks in the dataset
total_unique_tracks = spotify_data['track_name'].nunique()
total_unique_tracks
```

```python
# percentage of one hit wonders in the dataset
percentage_one_hit_wonders = (one_hit_wonder_tracks.shape[0] / total_unique_tracks) * 100
percentage_one_hit_wonders
```
