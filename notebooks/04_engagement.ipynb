```python
import pandas as pd

df = pd.read_parquet('../data/processed/spotify_history_clean.parquet')

df.head()
```

```python
# building a proxy for track length since it is not provided
# assumming the longest recordedplay of a track is the closest to its full length
df.groupby(['track_name', 'artist_name'])['ms_played'].max()
```

```python
# find completion rate per play

# creating a new column for the approx track length using transform and max
df['prox_track_length'] = df.groupby(['track_name', 'artist_name'])['ms_played'].transform('max')


# finding rate track was played by dividing the ms_played and its estimated length. 1 or 100% means it was played completely
df['rate_of_play'] = ( df['ms_played'] / df['prox_track_length'] )

df.head(20)

# all the rows of tracks not played meaning ms_played = 0 thus their rate is NaN
null_rows = df[df['rate_of_play'].isna()]

null_rows['ms_played'].describe()
```

```python
# Completion rate over the years
```

```python
# completion rate by year by median due toour data being left skewed
print(df.groupby('year')['rate_of_play'].agg(['mean', 'median']))
```

```python
# shows that major songs played in 2017 were played through the forward button
df[df['year'] == 2017]['reason_start'].value_counts(normalize=True)
```

```python
top_20_artists = df['artist_name'].value_counts().head(20)
top_20_artists


top_20_df = df[df['artist_name'].isin(top_20_artists.index)]

top_20_df
```

```python
# stats of the completion rate of the top 20 artists
top_20_df.groupby('artist_name')['rate_of_play'].agg(['mean','median']).sort_values(by='median', ascending=False)
```
